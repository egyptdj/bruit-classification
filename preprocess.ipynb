```python
import os
import glob
import shutil
```

```python
DATADIR = '/Users/egyptdj/Library/CloudStorage/Dropbox/Academic/01_Research/33_BruitApp_2022/v5'
FINALDIR = '/Users/egyptdj/Library/CloudStorage/Dropbox/Academic/01_Research/33_BruitApp_2022/data_v5'
```

```python
fnames =[f for f in glob.glob(os.path.join(DATADIR, '**', '*.wav'))]
```

```python
for f in fnames:
    label = f.split('/')[-2]
    os.makedirs(os.path.join(FINALDIR, 'train', label), exist_ok=True)
    os.makedirs(os.path.join(FINALDIR, 'test', label), exist_ok=True)
    if len(f.split(' ')) >= 4 and len(f.split(' ')[2])>0:
        label = f.split('/')[-2]
        subject_id = f.split(' ')[2]
        if 2487793 <= int(subject_id) <= 10585535:
            split = 'train'
        else:
            split = 'test'
        i = 0
        while True:
            if os.path.exists(os.path.join(FINALDIR, split, label, f'{subject_id}_{i}.wav')):
                i += 1
            else:
                shutil.copyfile(f, os.path.join(FINALDIR, split, label, f'{subject_id}_{i}.wav'))
                break
    else:
        i = 0
        while True:
            if os.path.exists(os.path.join(FINALDIR, 'train', label, f'0000_{i}.wav')):
                i += 1
            else:
                shutil.copyfile(f, os.path.join(FINALDIR, 'train', label, f'0000_{i}.wav'))
                break
    
```

```python
for f in glob.glob(os.path.join(FINALDIR, 'test', '**', '*.wav')):
    label = f.split('/')[-2]
    os.makedirs(os.path.join(FINALDIR, 'val', label), exist_ok=True)
    if int(f.split('.')[-2][-1])%2 == 0:
        shutil.move(f, os.path.join(FINALDIR, 'val', label, f.split('/')[-1]))
```

## Rules
- 0000: Train
- 1~2487792: Val (_even) / Test (_odd)
- 2487793~10585535: Train
- 10585536~: Val (_even) / Test (_odd)

```python
SAMPLE_RATE = 16000
for f in glob.glob(os.path.join(FINALDIR, '**', '**', '*.wav')):
    if f.endswith('_16b.wav'):
        print(os.system(f'soxi {f}'))
    else:
        print(f)
        os.system(f'sox {f} -b 16 -r {SAMPLE_RATE} {f.replace(".wav","_16b.wav")}')
        os.remove(f)
```
